```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
# question 1

def generate_gaussian_kernel(seg,siz):
    x=np.arange(-siz,siz)
    y=np.arange(-siz,siz)
    array=np.zeros((siz,siz))
    for i in x:
        for j in y:
            first=1/(2*np.pi*(seg**2))
            array[i][j]=first*(np.exp(-1*((i**2)+(j**2))/(2*(seg**2))))
    
    return array
            
        
```

```python
# queston 2

def convolution(img,mask,size):
    r1,c1=img.shape[:2]
    # print(r1,"  ",c1)
    # r2,c2=mask.shape()
    result= np.zeros((r1,c1))
    for i in range(0,r1):
        for j in range(0,c1):
            if(i>=(r1-size)):
                break
            elif (j>=(c1-size)):
                break
            new=img[i:i+size,j:j+size]
            value=np.multiply(mask,new)
            new_value=np.sum(value)
            result[i+(size//2)-1][j+(size//2)-1]=new_value
            
    return result
                 
    
```

```python
#question 4

k1=(1/9)*np.ones((3,3))
k2=generate_gaussian_kernel(3,7)
k2
```
