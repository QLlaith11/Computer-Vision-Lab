```python
# question 2
import numpy as np

def sort2d(array):
    array = np.array(array)
    flat_array = array.flatten()
    r,c=array.shape

    n = len(flat_array)
    for i in range(n):
        for j in range(0, n-i-1):
            if flat_array[j] > flat_array[j+1]:
                flat_array[j], flat_array[j+1] = flat_array[j+1], flat_array[j]

    max_val = flat_array[-1]
    min_val = flat_array[0]
    
    mid_index = n // 2
    if n % 2 == 0:  
        median_val = (flat_array[mid_index - 1] + flat_array[mid_index]) / 2
    else: 
        median_val = flat_array[mid_index]
        
    # sorted_array=flat_array.reshape(r,c)
    
    return max_val, min_val, median_val
```

```python

array = [[3, 1, 4], [1, 5, 9], [2, 6, 5]]
[max_val, min_val, median_val] = sort2d(array)
print(f'\n Max: {max_val}, Min: {min_val}, Median: {median_val}')
```

```python
#question 3
def nonLinearFilter(img, size):
    r, c = img.shape[:2]
    pad = size // 2 
    
  
    Min = np.zeros((r, c))
    Max = np.zeros((r, c))
    Med = np.zeros((r, c))
    
    for i in range(pad, r - pad):
        for j in range(pad, c - pad):
         
            part = img[i - pad:i + pad + 1, j - pad:j + pad + 1]

            [max_val, min_val, median] = sort2d (part)
            
            Min[i, j] = max_val
            Max[i, j] = min_val
            Med[i, j] = median

    
    return Min,Max,Med



# def nonLinearFilter(image_array ,size_of_the_kernel):
#     r,c=image_array.shape[:2]
#     # print("r="+str(r),"c= ",c)
#     Max_array=np.zeros((r,c))
#     Median_array=np.zeros((r,c))
#     Min_array=np.zeros((r,c))
    
#     pad=size_of_the_kernel//2
    
#     for i in range(pad,r-pad):
#         for j in range(pad,c-pad):
             
#             new_array=image_array[i-pad: i+pad+1,j:j-pad:j+pad+1]
#             [max_val,min_val,median_val]=sort2d(new_array)
#             Max_array[i,j]=max_val
#             Min_array[i,j]=min_val
#             Median_array[i,j]=median_val
            
#     return Max_array,Min_array,Median_array
```
